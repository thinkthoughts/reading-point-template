## Initialize Repository Runtime

The notebook initializes the repository once. All later artifacts are generated from the notebook specification.


```python
from __future__ import annotations

import importlib
import os
import shutil
import subprocess
import sys
from pathlib import Path

REPOSITORY_URL = "https://github.com/thinkthoughts/sensors-becker.git"
COLAB_REPOSITORY_ROOT = Path("/content/sensors-becker")


def install_colab_repository() -> Path:
    if COLAB_REPOSITORY_ROOT.exists():
        shutil.rmtree(COLAB_REPOSITORY_ROOT)

    subprocess.run(
        ["git", "clone", "--depth", "1", REPOSITORY_URL, str(COLAB_REPOSITORY_ROOT)],
        check=True,
    )
    subprocess.run(
        [
            sys.executable,
            "-m",
            "pip",
            "install",
            "--quiet",
            "--editable",
            str(COLAB_REPOSITORY_ROOT),
        ],
        check=True,
    )

    src_directory = COLAB_REPOSITORY_ROOT / "src"
    if str(src_directory) not in sys.path:
        sys.path.insert(0, str(src_directory))

    importlib.invalidate_caches()
    os.chdir(COLAB_REPOSITORY_ROOT)
    return COLAB_REPOSITORY_ROOT


try:
    import sensors_becker
except ModuleNotFoundError:
    repository_root = install_colab_repository()
    import sensors_becker
else:
    repository_root = Path(sensors_becker.__file__).resolve().parents[2]

from sensors_becker import initialize_notebook

runtime = initialize_notebook(
    start=repository_root,
    environment=(
        "google-colab"
        if repository_root == COLAB_REPOSITORY_ROOT
        else "repository-runtime"
    ),
)
context = runtime.context
runtime.validate()

print(f"Environment: {runtime.environment}")
print(f"Repository root: {runtime.repository_root}")
print("Engineering context validation: PASSED")
```

## Admit Notebook Specification

This cell is the notebook's engineering source of truth. Titles, dialogue, supporting context, artifact names, alt text, manifest records, and verification all derive from this specification.


```python
from dataclasses import dataclass


@dataclass(frozen=True)
class DialogueArtifactSpec:
    order: str
    artifact_id: str
    concept: str
    title: str
    first_label: str
    second_label: str
    supporting_context: tuple[str, str]


NOTEBOOK_SPEC = {
    "notebook_id": NOTEBOOK_ID,
    "notebook_version": NOTEBOOK_VERSION,
    "engineering_scope": "engineering_stability",
    "engineering_object": "Microcalorimeter",
    "engineering_direction": "Toward next-generation microcalorimeters.",
    "footer": "Admissible generalizations trail leading specifications.",
    "renderer_version": "3",
    "dialogue": (
        DialogueArtifactSpec(
            order="A",
            artifact_id="06_A_engineering_stability_trail",
            concept="Engineering Stability",
            title="Engineering Stability Trail: Microcalorimeters",
            first_label="Engineering Refinement",
            second_label="Engineering Stability",
            supporting_context=(
                "Observed Contrast",
                "Leading Specification",
            ),
        ),
        DialogueArtifactSpec(
            order="B",
            artifact_id="06_B_engineering_continuity_trail",
            concept="Engineering Continuity",
            title="Engineering Continuity Trail: Microcalorimeters",
            first_label="Engineering Stability",
            second_label="Engineering Continuity",
            supporting_context=(
                "Engineering Refinement",
                "Leading Specification",
            ),
        ),
        DialogueArtifactSpec(
            order="C",
            artifact_id="06_C_engineering_consistency_trail",
            concept="Engineering Consistency",
            title="Engineering Consistency Trail: Microcalorimeters",
            first_label="Engineering Continuity",
            second_label="Engineering Consistency",
            supporting_context=(
                "Observed Behavior",
                "Measurement Record",
            ),
        ),
        DialogueArtifactSpec(
            order="D",
            artifact_id="06_D_engineering_reliability_trail",
            concept="Engineering Reliability",
            title="Engineering Reliability Trail: Microcalorimeters",
            first_label="Engineering Consistency",
            second_label="Engineering Reliability",
            supporting_context=(
                "Engineering Continuity",
                "Leading Specification",
            ),
        ),
    ),
}

assert tuple(item.order for item in NOTEBOOK_SPEC["dialogue"]) == ("A", "B", "C", "D")
assert len({item.artifact_id for item in NOTEBOOK_SPEC["dialogue"]}) == 4

NOTEBOOK_SPEC
```

## Admit Reading Point

This Reading Point records how the admitted NB_06 engineering dialogue becomes inheritable by the next notebook.


```python
READING_POINT = {
    "reading_point_id": "RP_06",
    "notebook_id": NOTEBOOK_ID,
    "inherited_engineering_dialogue": (
        "Engineering Refinement",
        "Engineering Stability",
        "Engineering Continuity",
        "Engineering Consistency",
        "Engineering Reliability",
    ),
    "engineering_objective": None,
    "engineering_statements": (),
    "observed_contrast": "inherited",
    "engineering_candidates": (),
    "admitted_engineering_dialogue": (
        "Engineering Refinement",
        "Engineering Stability",
        "Engineering Continuity",
        "Engineering Consistency",
        "Engineering Reliability",
    ),
    "inheritability": {
        "status": "admitted",
        "next_notebook": "NB_07_A_ENGINEERING_CAPABILITY",
    },
}

assert READING_POINT["inherited_engineering_dialogue"] == READING_POINT["admitted_engineering_dialogue"]
assert READING_POINT["engineering_objective"] is None
assert READING_POINT["engineering_statements"] == ()
assert READING_POINT["engineering_candidates"] == ()

READING_POINT
```

## Prepare Specification Renderer

One generic renderer realizes every admitted dialogue artifact. The notebook contains no artifact-specific render function.


```python
import json
import zipfile
from datetime import date
from hashlib import sha256

from IPython.display import Image, Markdown, display

from sensors_becker.dialogue_renderer import (
    DialogueFigure,
    DialogueNode,
    DialogueRelation,
    NotebookDialogueRenderer,
)

renderer = NotebookDialogueRenderer(
    figsize=(12, 8),
    dpi=180,
    validate_layout=True,
)


def sha256_for_path(path: Path) -> str:
    digest = sha256()
    with path.open("rb") as handle:
        for chunk in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(chunk)
    return digest.hexdigest()


def display_artifact(path: Path) -> None:
    if path.suffix.lower() == ".png":
        display(Image(filename=str(path)))
    elif path.suffix.lower() == ".md":
        display(Markdown(path.read_text(encoding="utf-8")))
    else:
        print(path.read_text(encoding="utf-8"))
    print(f"Artifact: {path.name}")


def build_figure(spec: DialogueArtifactSpec) -> DialogueFigure:
    return DialogueFigure(
        title=spec.title,
        subtitle=NOTEBOOK_SPEC["engineering_direction"],
        footer=NOTEBOOK_SPEC["footer"],
        primary_nodes=(
            DialogueNode(
                label=spec.first_label,
                x=0.5,
                y=0.67,
                width=0.42,
                height=0.10,
                role="input",
                fontsize=18,
            ),
            DialogueNode(
                label=spec.second_label,
                x=0.5,
                y=0.46,
                width=0.42,
                height=0.12,
                role="primary",
                emphasis=True,
                fontsize=21,
            ),
        ),
        primary_relations=(
            DialogueRelation(
                start=(0.5, 0.62),
                end=(0.5, 0.52),
                role="input",
                directional=True,
                linewidth=2.0,
            ),
        ),
        supporting_context=spec.supporting_context,
    )


def build_alt_text(spec: DialogueArtifactSpec) -> str:
    left, right = spec.supporting_context
    return (
        "# Alt Text\n\n"
        f"**{spec.title}.** Diagram showing {spec.first_label.lower()} "
        f"leading to {spec.second_label.lower()}. "
        f"{left} and {right} appear as supporting engineering context. "
        f"Subtitle: “{NOTEBOOK_SPEC['engineering_direction']}” "
        f"Footer: “{NOTEBOOK_SPEC['footer']}”\n"
    )


bundle_directory = runtime.paths.outputs / "releases" / NOTEBOOK_ID
bundle_directory.mkdir(parents=True, exist_ok=True)

for prior_path in bundle_directory.iterdir():
    if prior_path.is_file():
        prior_path.unlink()

print(f"Renderer: {renderer.__class__.__name__}")
print("Architecture: specification-driven")
print(f"Artifact specifications: {len(NOTEBOOK_SPEC['dialogue'])}")
print(f"Bundle directory: {runtime.relative_path(bundle_directory)}")
```

## Generate Admitted Dialogue Bundle

The following loop renders A–D, writes matching alt text, displays each realization, and records every generated path.


```python
generated_records = []

for artifact_spec in NOTEBOOK_SPEC["dialogue"]:
    png_path = bundle_directory / f"{artifact_spec.artifact_id}.png"
    alt_path = bundle_directory / f"{artifact_spec.artifact_id}.alt.md"

    renderer.render(build_figure(artifact_spec), png_path)
    alt_path.write_text(build_alt_text(artifact_spec), encoding="utf-8")

    generated_records.append(
        {
            "spec": artifact_spec,
            "png_path": png_path,
            "alt_path": alt_path,
        }
    )

    display_artifact(png_path)
    display_artifact(alt_path)

assert len(generated_records) == len(NOTEBOOK_SPEC["dialogue"])
```

## Record Portable Bundle

README, metadata, and manifest records are generated from the same notebook specification.


```python
readme_path = bundle_directory / "06_E_README.md"
metadata_path = bundle_directory / "06_F_notebook_metadata.json"
manifest_path = bundle_directory / "06_G_manifest.json"

dialogue_lines = "\n\n↓\n\n".join(
    item.concept for item in NOTEBOOK_SPEC["dialogue"]
)

reading_point_lines = "\n\n↓\n\n".join(
    READING_POINT["admitted_engineering_dialogue"]
)

artifact_lines = "\n".join(
    f"- {item.order}: {item.artifact_id}"
    for item in NOTEBOOK_SPEC["dialogue"]
)

readme_text = f"""# {NOTEBOOK_ID}

This bundle contains the notebook's admitted engineering concepts in portable reading order.

## Engineering Dialogue

{dialogue_lines}

## Reading Point

Inherited Engineering Dialogue

{reading_point_lines}

Engineering Objective: pending next Reading Point

Engineering Statements: none admitted in NB_06

Observed Contrast: inherited

Engineering Candidates: none admitted in NB_06

Admitted Engineering Dialogue

{reading_point_lines}

## Engineering Artifacts

{artifact_lines}

## Repository

{context.repository}

## Engineering Object

{NOTEBOOK_SPEC["engineering_object"]}

## Engineering Direction

{NOTEBOOK_SPEC["engineering_direction"]}

---

*{NOTEBOOK_SPEC["footer"]}*
"""

readme_path.write_text(readme_text, encoding="utf-8")

metadata = {
    "artifact_type": "notebook_bundle",
    "notebook_id": NOTEBOOK_ID,
    "notebook_filename": NOTEBOOK_FILENAME,
    "notebook_version": NOTEBOOK_VERSION,
    "repository": context.repository,
    "engineering_stage": ENGINEERING_STAGE,
    "engineering_scope": NOTEBOOK_SPEC["engineering_scope"],
    "engineering_object": NOTEBOOK_SPEC["engineering_object"],
    "engineering_direction": NOTEBOOK_SPEC["engineering_direction"],
    "renderer": {
        "name": renderer.__class__.__name__,
        "version": NOTEBOOK_SPEC["renderer_version"],
    },
    "architecture": "specification-driven",
    "reading_point": {
        "reading_point_id": READING_POINT["reading_point_id"],
        "inherited_engineering_dialogue": list(READING_POINT["inherited_engineering_dialogue"]),
        "engineering_objective": READING_POINT["engineering_objective"],
        "engineering_statements": list(READING_POINT["engineering_statements"]),
        "observed_contrast": READING_POINT["observed_contrast"],
        "engineering_candidates": list(READING_POINT["engineering_candidates"]),
        "admitted_engineering_dialogue": list(READING_POINT["admitted_engineering_dialogue"]),
        "inheritability": READING_POINT["inheritability"],
    },
    "runtime_environment": runtime.environment,
    "generated": date.today().isoformat(),
}

metadata_path.write_text(
    json.dumps(metadata, indent=2) + "\n",
    encoding="utf-8",
)

manifest_artifacts = []

for record in generated_records:
    artifact_spec = record["spec"]
    for realization_order, path in enumerate(
        (record["png_path"], record["alt_path"]),
        start=1,
    ):
        manifest_artifacts.append(
            {
                "artifact_order": artifact_spec.order,
                "artifact_id": artifact_spec.artifact_id,
                "engineering_concept": artifact_spec.concept,
                "primary_dialogue": [
                    artifact_spec.first_label,
                    artifact_spec.second_label,
                ],
                "supporting_context": list(artifact_spec.supporting_context),
                "realization_order": realization_order,
                "realization_type": (
                    "png" if path.suffix.lower() == ".png" else "alt_text"
                ),
                "filename": path.name,
                "size_bytes": path.stat().st_size,
                "sha256": sha256_for_path(path),
                "status": "verified",
            }
        )

for order, path in (("E", readme_path), ("F", metadata_path)):
    manifest_artifacts.append(
        {
            "artifact_order": order,
            "artifact_id": path.stem,
            "engineering_concept": "bundle_record",
            "realization_order": 1,
            "realization_type": path.suffix.lower().lstrip("."),
            "filename": path.name,
            "size_bytes": path.stat().st_size,
            "sha256": sha256_for_path(path),
            "status": "verified",
        }
    )

manifest = {
    "release_filename": RELEASE_FILENAME,
    "repository": context.repository,
    "notebook_id": NOTEBOOK_ID,
    "notebook_version": NOTEBOOK_VERSION,
    "engineering_stage": ENGINEERING_STAGE,
    "architecture": "specification-driven",
    "reading_point": {
        "reading_point_id": READING_POINT["reading_point_id"],
        "inherited_engineering_dialogue": list(READING_POINT["inherited_engineering_dialogue"]),
        "engineering_objective": READING_POINT["engineering_objective"],
        "engineering_statements": list(READING_POINT["engineering_statements"]),
        "observed_contrast": READING_POINT["observed_contrast"],
        "engineering_candidates": list(READING_POINT["engineering_candidates"]),
        "admitted_engineering_dialogue": list(READING_POINT["admitted_engineering_dialogue"]),
        "inheritability": READING_POINT["inheritability"],
    },
    "renderer_version": NOTEBOOK_SPEC["renderer_version"],
    "generated": date.today().isoformat(),
    "engineering_dialogue": [
        item.concept for item in NOTEBOOK_SPEC["dialogue"]
    ],
    "artifacts": manifest_artifacts,
}

manifest_path.write_text(
    json.dumps(manifest, indent=2) + "\n",
    encoding="utf-8",
)

for path in (readme_path, metadata_path, manifest_path):
    display_artifact(path)
```

## Verify Specification and Bundle

Verification compares the generated bundle back to the admitted specification rather than to manually repeated notebook code.


```python
artifact_paths = [
    path
    for record in generated_records
    for path in (record["png_path"], record["alt_path"])
]
bundle_artifact_paths = artifact_paths + [
    readme_path,
    metadata_path,
    manifest_path,
]

expected_names = [
    name
    for item in NOTEBOOK_SPEC["dialogue"]
    for name in (
        f"{item.artifact_id}.png",
        f"{item.artifact_id}.alt.md",
    )
] + [
    "06_E_README.md",
    "06_F_notebook_metadata.json",
    "06_G_manifest.json",
]

actual_names = [path.name for path in bundle_artifact_paths]

assert actual_names == expected_names
assert all(path.exists() and path.stat().st_size > 0 for path in bundle_artifact_paths)

with metadata_path.open(encoding="utf-8") as handle:
    metadata_check = json.load(handle)

with manifest_path.open(encoding="utf-8") as handle:
    manifest_check = json.load(handle)

assert metadata_check["architecture"] == "specification-driven"
assert manifest_check["architecture"] == "specification-driven"
assert metadata_check["reading_point"]["reading_point_id"] == "RP_06"
assert metadata_check["reading_point"]["engineering_objective"] is None
assert metadata_check["reading_point"]["engineering_statements"] == []
assert metadata_check["reading_point"]["engineering_candidates"] == []
assert metadata_check["reading_point"]["observed_contrast"] == "inherited"
assert metadata_check["reading_point"]["inheritability"]["status"] == "admitted"
assert manifest_check["reading_point"] == metadata_check["reading_point"]

assert manifest_check["engineering_dialogue"] == [
    item.concept for item in NOTEBOOK_SPEC["dialogue"]
]

for record in manifest_check["artifacts"]:
    path = bundle_directory / record["filename"]
    assert path.exists()
    assert sha256_for_path(path) == record["sha256"]

for item, generated in zip(NOTEBOOK_SPEC["dialogue"], generated_records):
    assert generated["png_path"].stem == item.artifact_id
    assert generated["alt_path"].name == f"{item.artifact_id}.alt.md"

for path in bundle_artifact_paths:
    print(f"✓ {path.name} ({path.stat().st_size} bytes)")

print("✓ Notebook specification verified")
print("✓ Generated artifacts match the admitted dialogue")
print("✓ Specification-driven architecture verified")
print("✓ Reading Point architecture verified")
print("✓ Inherited engineering dialogue prepared for NB_07")
```

## Package and Download

The verified portable bundle is packaged in specification order.


```python
release_directory = runtime.paths.outputs / "releases"
release_directory.mkdir(parents=True, exist_ok=True)
release_path = release_directory / RELEASE_FILENAME

with zipfile.ZipFile(release_path, "w", zipfile.ZIP_DEFLATED) as archive:
    for path in bundle_artifact_paths:
        archive.write(path, arcname=path.name)

with zipfile.ZipFile(release_path) as archive:
    assert archive.namelist() == expected_names

print(
    f"✓ {runtime.relative_path(release_path)} "
    f"({release_path.stat().st_size} bytes)"
)

if runtime.environment == "google-colab":
    from google.colab import files
    files.download(str(release_path))
else:
    print(release_path)
```
